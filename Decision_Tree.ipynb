```python
import pandas as pd
df = pd.read_csv("salaries.csv")
df.head()
```

```python
inputes = df.drop('salary_more_then_100k',axis='columns')
target = df['salary_more_then_100k']
```

```python
from sklearn.preprocessing import LabelEncoder
le_company = LabelEncoder()
le_job = LabelEncoder()
le_degree = LabelEncoder()
```

```python
inputes['company_n'] = le_company.fit_transform(inputes['company'])
inputes['job_n'] = le_job.fit_transform(inputes['job'])
inputes['degree_n'] = le_degree.fit_transform(inputes['degree'])
inputes.head()
```

```python
inputes_n = inputes.drop(['company','job','degree'],axis='columns')
inputes_n
```

```python
from sklearn import tree
model = tree.DecisionTreeClassifier()
```

```python
model.fit(inputes_n, target)
```

```python
model.score(inputes_n,target)
```

```python
import pandas as pd

# Create a DataFrame for prediction with the same column names as inputes_n
prediction_input = pd.DataFrame([[2, 2, 1]], columns=inputes_n.columns)

model.predict(prediction_input)
```

```python
model.predict([[1,1,1]])
```

```python
#Exercise
import pandas as pd
df = pd.read_csv("titanic.csv")
df.head()
```

```python
df.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns',inplace=True)
df.head()
```

```python
inputs = df.drop('Survived',axis='columns')
target = df.Survived
```

```python
inputs.Sex = inputs.Sex.map({'male':1,'female':2})
```

```python
inputs.Age[:10]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(inputs,target,test_size=0.2)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn import tree
model = tree.DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
model.predict(x_test)
```

```python
model.score(x_test,y_test)
```

```python
model.predict([[3,2,	26.0,	7.9250]])
```
